# **Coding Questions (no for-loop)**

**1.	Create an empty tuple.**



```python
empty_tuple = ()

print(empty_tuple)
```

**2.	Create a tuple with 5 integers.**


```python
my_tuple = (3, 6, 7, 10, 15)

print(my_tuple)
```

**3.	Access 3rd element of a tuple.**


```python
my_tuple = (3, 6, 7, 10, 15)

print(my_tuple[2])
```

**4.	Access last element without negative index.**


```python
my_tuple = (3, 6, 7, 10, 15)

l = len(my_tuple)

print(my_tuple[l-1])
```

**5.	Slice first three elements.**

```python
my_tuple = (3, 6, 7, 10, 15)

print(my_tuple[:3])
```

**6.	Slice every alternate value.**


```python
my_tuple = (3, 6, 7, 10, 15, 100, 99)

print(my_tuple[::2])
```

**7.	Reverse a tuple using slicing.**


```python
my_tuple = (3, 6, 7, 10, 15, 99, 100)

print(my_tuple[::-1])
```

**8.	Concatenate two tuples.**


```python
tu1 = (3, 6, 7)
tu2 = (10, 15, 99, 100)

print(tu1 + tu2)
```

**9.	Replicate tuple (1,2) twice.**


```python
my_tuple = (1, 2)
replicated_tuple = my_tuple * 2

print(replicated_tuple)
```

**10.	Find length of a tuple.**

```python
my_tuple = (3, 6, 7, 10, 15, 99, 100)

print(len(my_tuple))
```


**11.	Convert a list to a tuple.**



```python
my_list  = [1, 2, 'ape', 6.5]

print(tuple(my_list))
```

**12.	Convert a string "python" to tuple of characters.**


```python
s = 'python'

print(tuple(s))
```

**13.	Check if value exists in tuple.**


```python
tu = (3, 6, 7, 10, 15, 99, 100)


a = int(input())

if a in tu:
  print(f"{a} exists in given tuple")
else:
  print(f"{a} not exists in given tuple")
```

**14.	Find maximum element in tuple.**

```python
tu = (3, 6, 7, 10, 15, 99, 100)

print(max(tu))
```

**15.	Find minimum element in tuple.**


```python
tu = (3, 6, 7, 10, 15, 99, 100)

print(min(tu))
```

**16.	Count how many times 2 appears in (1,2,2,3).**


```python
gt = (1, 2, 2, 3)

print(gt.count(2))
```

**17.	Find index of an element.**


```python
gt = (1, 2, 2, 6, 3)

gt.index(6)
```

**18.	Perform tuple packing.**


```python
a = 10
b = 20
c = 30

packed_tuple = a, b, c

print(packed_tuple)
```

**19.	Perform tuple unpacking into variables.**


```python
a, b, c = (1, 2, 3) #unpacking the tuple (1, 2, 3) & assigning the values to variables to a, b, c

print(a)
print(b)
print(c)
```

**20.	Convert tuple to string without loops.**

```python
my_tuple = ('H', 'e', 'l', 'l', 'o')

# Convert tuple of characters to a string using join()
my_string = "".join(my_tuple)

print(my_string)
```
