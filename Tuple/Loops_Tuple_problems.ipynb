## **For loop with tuple:**


*   Used to iterate over and process each element in a tuple one by one.


**1.	Write a for loop program to print all elements of a tuple.**


```python
my_tuple = (13, 11, 64, 96, 101)

for num in my_tuple:
  print(num, end = " ")
```

**3.	Write a for loop program to count how many even numbers are present in a tuple.**


```python
my_tuple = (13, 11, 64, 96, 101)

count = 0
for num in my_tuple:
  if num % 2 == 0:
    count += 1

print(f"Count of even numbers in my tuple: {count}")
```

**5. Write a for loop program to print the index and value of each tuple element.**

```python
my_tuple = (13, 11, 64, 96, 101)

for idx, num in enumerate(my_tuple):
  print(idx, num)
```

**7. Write a for loop program to create a new tuple containing only odd numbers from an existing tuple.**

```python
my_tuple = tuple(map(int, input("Enter tuple elements separated by spaces: ").split()))

odd_list = []

for num in my_tuple:
  if num % 2 != 0:
    odd_list.append(num)

print(tuple(odd_list))
```

**9. Write a for loop program to calculate the sum of all numbers inside a tuple.**

```python
my_tuple = (13, 11, 64, 96, 101)

sum = 0
for num in my_tuple:
  sum += num

print(sum)
```

**11. Write a for loop program to find how many strings in a tuple have length greater than 5.**

```python
my_string_tuple = ("apple", "banana", "cherry", "date", "elderberry")

count = 0
for word in my_string_tuple:
  if len(word) > 5:
    count += 1
print(f"No. of strings in tuple having length > 5 : {count}")
```

**13. Write a for loop program to separate elements of a tuple into two new tuples: one with even numbers and one with odd numbers.**

```python
my_tuple = tuple(map(int, input("Enter tuple elements separated by spaces: ").split()))

odd_list = []
even_list = []
for num in my_tuple:
  if num % 2 != 0:
    odd_list.append(num)
  else:
    even_list.append(num)

print(tuple(odd_list))
print(tuple(even_list))
```

**15. Write a for loop program to print only those tuple elements that are divisible by 3.**

```python
my_tuple = tuple(map(int, input("Enter tuple elements separated by spaces: ").split()))


for num in my_tuple:
  if num % 3 == 0:
    print(num)
```

# **While loop with tuple:**

*    Used to repeatedly execute a block of code as long as a specified condition related to the tuple (often an index) remains true.

**2.	Write a while loop program to print all elements of a tuple using indexes.**


```python
my_tuple = tuple(map(int, input("Enter tuple elements separated by spaces: ").split()))

L = len(my_tuple)

i = 0
while i < L:
  print(my_tuple[i])
  i += 1
```

**4. Write a while loop program to find the largest value in a tuple.**

```python
my_tuple = tuple(map(int, input("Enter tuple elements separated by spaces: ").split()))

L = len(my_tuple)

i = 0
max_val = my_tuple[0]
while i < L:
  if my_tuple[i] > max_val:
    max_val = my_tuple[i]
  i += 1
print(max_val)
```

**6. Write a while loop program to reverse a tuple manually without using slicing.**

```python
my_tuple = tuple(map(int, input("Enter tuple elements separated by spaces: ").split()))

L = len(my_tuple)

rev_list = []

i = 0
while i < L:
  rev_list.append(my_tuple[L-i-1])
  i += 1
print(tuple(rev_list))
```

**8. Write a while loop program that prints elements until a specific value is found, then stops.**

```python
my_tuple = tuple(map(int, input("Enter tuple elements separated by spaces: ").split()))

L = len(my_tuple)

n = int(input())

i = 0
while i < L:
  print(my_tuple[i])
  if my_tuple[i] == n:
    break
  i = i + 1
```

**10. Write a while loop program to count how many times a particular element occurs in a tuple.**

```python
my_tuple = tuple(map(int, input("Enter tuple elements separated by spaces: ").split()))

L = len(my_tuple)

i = 0
count_dict = {}
while i < L:
  if my_tuple[i] not in count_dict:
    count_dict[my_tuple[i]] = 1
  else:
    count_dict[my_tuple[i]] += 1
  i = i + 1
print(count_dict)
```

**12. Write a while loop program to print alternate elements of a tuple.**

```python
my_tuple = tuple(map(int, input("Enter tuple elements separated by spaces: ").split()))

L = len(my_tuple)

i = 0
while i < L:
  print(my_tuple[i])
  i += 2
```

**14. Write a while loop program to copy elements from one tuple to a new tuple.**

```python
my_tuple = tuple(map(int, input("Enter tuple elements separated by spaces: ").split()))

L = len(my_tuple)

list_copy = []

i = 0
while i < L:
  list_copy.append(my_tuple[i])
  i += 1
print(tuple(list_copy))
```
