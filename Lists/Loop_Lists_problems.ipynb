## **For loop with lists:**


*   Used to iterate over and process each element in a list one by one.


**1.	Write a for loop program to print each element of a list on a new line.**


```python
list_a = ["Six", 4, 6.4, True]

for item in list_a:
  print(item)
```

**2.	Write a for loop program to find the sum of all elements in a list.**


```python
list_a = [5, 45, 64, 100]

sum = 0
for num in list_a:
  sum += num

print(sum)
```

**3.	Write a for loop program to print only even numbers from a list.**


```python
list_a = [5, 45, 64, 100]

for num in list_a:
  if (num % 2 == 0):
    print(f"{num} is even")
```

**4.	Write a for loop program to count how many times a specific element appears in a list.**


```python
list_a = [5, 45, 64, 45, 64, 100, 45]


n = int(input())
c = 0
for num in list_a:
  if num == n:
    c += 1

print(f"Given number {n} is repeated {c} times")
```

**5.	Write a for loop program to print the index and value of each element.**


```python
list_a = ["six", 4, 6.4, True]

for index, item in enumerate(list_a):
  print (index, item)
```

**6.	Write a for loop program to print elements greater than a given number.**

```python
n = int(input())  #given number

list_a = [5, 45, 64, 100]

for val in list_a:
  if val > n:
    print(val)
```

**7.	Write a for loop program that multiplies each element by 3 and prints the updated values.**


```python
list_a = [5, 45, 64, 100]

for num in list_a:
  print(num * 3)
```

**8.	Write a for loop program to separate elements into two new lists: one with even numbers and one with odd numbers.**


```python
list_a = [5, 28, 45, 64, 81, 100]

even_list = []
odd_list = []

for num in list_a:
  if num % 2 == 0:
    even_list.append(num)
  else:
    odd_list.append(num)

print(even_list)
print(odd_list)
```

# **While loop with lists:**

*    Used to repeatedly execute a block of code as long as a specified condition related to the list (often an index) remains true.

**1.	Write a while loop program to print all items of a list using indexing.**


```python
list_a = [5, 33, 45, 64, 100]

L = len(list_a)

i = 0
while i < L:
  print(list_a[i])
  i += 1
```

**2.	Write a while loop program to find the largest number in a list.**


```python
list_a = [5, 33, 45, 64, 100]

max_val = 0

i = 0
while i < len(list_a): # Iterate through all elements of the list
  if list_a[i] > max_val:
    max_val = list_a[i]
  i += 1

print(max_val)
```

**3.	Write a while loop program that creates a new list containing only odd numbers from the original list.**


```python
list_a = [5, 33, 45, 64, 100]

i = 0
list_odd = []
while i < len(list_a):
  if list_a[i] % 2 != 0:
    list_odd.append(list_a[i])
  i = i + 1

print(list_odd)
```

**4.	Write a while loop program to reverse a list without using the reverse function.**


```python
list_a = [5, 33, 45, 64, 100]
l = len(list_a)

rev_list_a = []

i = 0
while i < l:
  rev_list_a.append(list_a[l-i-1])
  i += 1

print(rev_list_a)
```

**5.	Write a while loop program to copy all elements of a list into a new list.**


**6.	Write a while loop program that stops when it encounters the number 0 in a list.**

```python
list_a = [5, 33, 45, 64, 0, 100]

i = 0
while i < len(list_a):
  print(list_a[i])
  if list_a[i] == 0 :
    break

  i += 1
```

**7.	Write a while loop program to count how many integers in the list are divisible by 5.**


```python
list_a = [5, 33, 45, 64, 0, 100]

i = 0
c = 0
while i < len(list_a):
  if list_a[i] % 5 == 0:
    c = c + 1
  i = i + 1

print(f"count is {c}")
```
