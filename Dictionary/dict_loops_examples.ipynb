##**For Loops in Dictionary:**

A `for` loop can be used to iterate over the contents of a dictionary. By default, when you iterate over a dictionary directly, it iterates through its **keys**. You can also explicitly iterate over:

*   **Keys only**: `for key in my_dict.keys():`
*   **Values only**: `for value in my_dict.values():`
*   **Key-value pairs (items)**: `for key, value in my_dict.items():`

**1. Write a for loop to print all keys of a dictionary.**

```python
product_details = {
    'product_id': 'SKU001',
    'name': 'Wireless Ergonomic Mouse',
    'category': 'Computer Peripherals',
    'price': 29.99,
    'in_stock': True,
    'ratings': {'average': 4.5, 'count': 120}
}

for key, value in product_details.items():
  print(key)
```

**2. Write a for loop to print all values of a dictionary.**

**4. Write a program to count how many values in a dictionary are integers using a for loop.**

```python
groceries_list = {
    "apples": 10,        # int
    "oranges": 5,       # int
    "milk_liters": 2.5, # float
    "price": 99.75,     # float
    "rating": 4.2       # float
}

values = list(groceries_list.values())

int_count = 0
for val in values:
  if type(val) == int:
    int_count += 1

print(f"count of integer values in the dictionary are {int_count}")
```

**6. Write a for loop to print keys whose values are greater than a given number.**

## **While Loops in Dictionary**

*   A `while` loop can also be used to iterate over dictionary contents, though it's less common than `for` loops as it requires manual index management.
*   To use a `while` loop, you typically convert dictionary keys or items into a list and then iterate over that list using an index.

**Convert to list**: `keys_list = list(my_dict.keys())` or `items_list = list(my_dict.items())`

**Manual indexing**: Use an index variable (`i`) that is incremented in each iteration and checked against the length of the list.

**1. Using a while loop, iterate through dictionary keys after converting them into a list.**

```python
prod_details = {
    'product_id': 'SKU001',
    'name': 'Lenovo Wireless Ergonomic Mouse',
    'category': 'Computer Peripherals',
    'price': 29.99,
    'in_stock': True,
    'ratings': [3.5, 4, 4.5]
}

dict_keys = list(prod_details.keys())
print(dict_keys)

i = 0
while i < len(dict_keys):
  print(dict_keys[i])
  i += 1
```

**2. Write a while loop to print dictionary items until all keys are printed.**

```python
groceries_list = {
    "apples": 10,        # int
    "oranges": 5,       # int
    "milk_liters": 2.5, # float
    "price": 99.75,     # float
    "rating": 4.2       # float
}

keys = list(groceries_list.keys())

i = 0
while i < len(keys):
  print(keys[i])
  i += 1
```

**3. Write a program to reverse iterate over dictionary keys using a while loop.**

```python
groceries_list = {
    "apples": 10,        # int
    "oranges": 5,       # int
    "milk_liters": 2.5, # float
    "price": 99.75,     # float
    "rating": 4.2       # float
}

keys = list(groceries_list.keys())

i = len(keys) - 1
while i >= 0:
  print(keys[i])
  i -= 1
```

**4. Write a program using while loop to print alternate dictionary keys.**

```python
groceries_list = {
    "apples": 10,        # int
    "oranges": 5,       # int
    "milk_liters": 2.5, # float
    "price": 99.75,     # float
    "rating": 4.2       # float
}

keys = list(groceries_list.keys())

i = 0
while i < len(keys):
  print(keys[i])
  i += 2
```

**5. Write a while loop program to iterate through dictionary items by manually tracking an index.**

```python
groceries_list = {
    "apples": 10,        # int
    "oranges": 5,       # int
    "milk_liters": 2.5, # float
    "price": 99.75,     # float
    "rating": 4.2       # float
}

items = list(groceries_list.items())

print(items)
print(len(items))

i = 0
while i < len(items):
  print(items[i])
  i += 1
```
